#maximum marginal retrieval


```python
from langchain_core.documents import Document
```

```python
documents = [
    Document(page_content="Langchain helps developers build LLM applications easily"),
    Document(page_content="Chroma is an open-source vector database used to store and retrieve embeddings efficiently"),
    Document(page_content="Embeddings are numerical vector representations of text that capture semantic meaning for similarity search"),
    Document(page_content="Hugging Face provides pre-trained transformer models and datasets for natural language processing tasks"),
    Document(page_content="LangChain provides chains and agents to connect LLMs with tools, memory, and external data sources"),
    Document(page_content="LangChain supports multiple LLM providers including OpenAI, Anthropic, Cohere, and open-source models via Hugging Face"),
    Document(page_content="Retrieval-Augmented Generation (RAG) is a technique in LangChain that combines document retrieval with LLM generation"),
    Document(page_content="LangChain memory modules allow conversational agents to retain context across multiple interactions"),
    Document(page_content="LangChain document loaders can ingest data from PDFs, web pages, databases, and APIs for processing"),
    Document(page_content="LangSmith is a debugging and monitoring platform built alongside LangChain for tracing LLM application behavior"),
]
```

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)


vectorstore = FAISS.from_documents(
    documents=documents,
    embedding=embedding_model
)
```

```python
retriever = vectorstore.as_retriever(
    search_type= "mmr",
    search_kwargs= {"k" : 3, "lambda_mult": 0.5}#k=top results, lambda_mult = relevance-diversity balance
)
```

```python
query = "What is a langchain?"
results = retriever.invoke(query)
```

```python
for i, doc in enumerate(results) :
    print(f"\n--- Result {i+1}----") 
    print(doc.page_content)
```
