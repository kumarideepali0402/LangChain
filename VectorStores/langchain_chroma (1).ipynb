```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
from langchain_core.documents import Document

doc1 = Document(
    page_content="Virat Kohli is one of the greatest batsmen of his generation, known for his aggressive batting style and exceptional fitness. He has scored over 26,000 international runs across all formats.",
    metadata={"name": "Virat Kohli", "team": "Royal Challengers Bengaluru"}
)

doc2 = Document(
    page_content="Rohit Sharma is renowned for his elegant stroke play and is the only batsman to score three double centuries in ODIs. He is also the current captain of the Indian national team.",
    metadata={"name": "Rohit Sharma", "team": "Mumbai Indians"}
)

doc3 = Document(
    page_content="MS Dhoni is a legendary wicket-keeper batsman famous for his calm demeanor under pressure and iconic helicopter shot. He led India to three ICC trophies including the 2011 World Cup.",
    metadata={"name": "MS Dhoni", "team": "Chennai Super Kings"}
)

doc4 = Document(
    page_content="Hardik Pandya is a dynamic all-rounder who can change the game with both bat and ball. He played a crucial role in India's 2022 T20 World Cup campaign with his match-winning performances.",
    metadata={"name": "Hardik Pandya", "team": "Mumbai Indians"}
)

doc5 = Document(
    page_content="Jasprit Bumrah is considered one of the best fast bowlers in the world, known for his unorthodox action and deadly yorkers. He has been ranked number one in ICC Test bowling rankings.",
    metadata={"name": "Jasprit Bumrah", "team": "Mumbai Indians"}
)

doc6 = Document(
    page_content="Ravindra Jadeja is a valuable all-rounder known for his left-arm spin bowling, explosive lower-order batting, and brilliant fielding. He is consistently ranked among the top all-rounders in Test cricket.",
    metadata={"name": "Ravindra Jadeja", "team": "Chennai Super Kings"}
)
```

```python
docs = [doc1, doc2, doc3, doc4, doc5, doc6]
```

```python
vector_store = Chroma(
    embedding_function=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2"),
    persist_directory='my_chroma_db',
    collection_name='sample'
)
```

```python
ids = vector_store.add_documents(docs)
```

```python
vector_store.get(include=["embeddings", "documents", "metadatas"])
```

```python
vector_store.similarity_search(
    query='Who among these is a bowler',
    k=2 #no. of obj to show
)
```

```python
vector_store.similarity_search_with_score(
   query='Who among these is a bowler',
    k=2 #no. of obj to show
)
```

```python
vector_store.similarity_search_with_score(
    query="",
    filter={"team":"Chennai Super Kings"}
) #shows a score along with each result lesser the score more similarity
```

```python
updated_doc1 = Document(
    page_content="Virat Kohli is one of the greatest batsmen of his generation, known for his aggressive batting style and exceptional fitness. He has scored over 26,000 international runs across all formats.",
    metadata={'team': 'Royal Challengers Bangalore'}
)

vector_store.update_document(document_id=ids[0], document=updated_doc1)
```

```python
#view doc
vector_store.get(include=["embeddings", "documents", "metadatas"])
```

```python
#delete documents by their IDs (using the IDs returned from add_documents)
vector_store.delete(ids=[ids[0], ids[1]])
```
